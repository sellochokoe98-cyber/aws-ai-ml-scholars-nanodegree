```python
# Scatterplot showing negative correlation between two variables

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sb

fuel_econ = pd.read_csv(r'C:\Users\sello\Downloads\fuel-econ.csv')
plt.scatter(data=fuel_econ, x='displ', y='comb');
plt.xlabel('Displacement (1)')
plt.ylabel('Combined fuel Eff. (mpg)')
```

```python
# Plot the regression line on the transformed data

def log_trans(x, inverse=False):
    if not inverse:
        return np.log10(x)
    else:
        return np.power(10, x)

sb.regplot(data=fuel_econ, x=fuel_econ['displ'], y=fuel_econ['comb'].apply(log_trans))
tick_locs = [10, 20, 50, 100]
plt.yticks(log_trans(tick_locs), tick_locs);
```
