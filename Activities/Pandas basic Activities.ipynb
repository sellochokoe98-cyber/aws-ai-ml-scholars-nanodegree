# *Let's use a Pandas Series to store a grocery list.*

```python
import pandas as pd

#Pandas Series that stores a grocery list
groceries = pd.Series(data = [30, 6, 'Yes', 'No'], index = ['eggs', 'apples', 'milk', 'bread'])

print(groceries)
```

```python
# Print attributes -shape, ndim, and size

# We print some information about Groceries
print('Groceries has shape:', groceries.shape)
print('Groceries has dimension:', groceries.ndim)
print('Groceries has a total of', groceries.size, 'elements')
```

# *Accessing and Deleting Elements in pandas Series*

```python
# Access elemnts in Groceries using index labels:

# Single index label
print('How many eggs do we need to buy:', groceries['eggs'])
print()

# Access multiple index labels
print('Do we need milk and bread:\n', groceries[['milk', 'bread']])
print()

# loc to access multiple  index labels
print('How many eggs and apples do we need to buy:\n', groceries.loc[['eggs', 'apples']])
print()

# Access elements in Groceries using numerical indices:

# Multiple numerical indices
print('How many eggs and apples do we need to buy:\n', groceries.iloc[[0, 1]])
print

# Negative numerical index
print('Do we need bread:\n', groceries.iloc[[-1]])
print()

# Single numerical index
print('How many eggs do we need to buy:', groceries.iloc[0])
print()
# Single numerical index
print('Do we need milk and bread:\n', groceries.iloc[[2, 3]])
```

```python
# Delete elements in-place

# Display the origina grocery list
print('Original Grocery List:\n', groceries)

# Remove apples from our grocery list in place by setting the inplace keyword to True
groceries.drop('apples', inplace = True)

# Remove elements in place the original Series its modified. To see this
# Display our grocery list again
print()
print('Grocery List after removing apples in place:\n', groceries)
```

# *Arthmetic Operations on Pandas Series*

```python
# Create a Pandas Series that stores a grocery list of just fruits
fruits= pd.Series(data = [10, 6, 3,], index = ['apples', 'oranges', 'bananas'])

# Display the fruits Pandas Series
fruits
```

```python
# Print fruits for reference
print('Original grocery list of fruits:\n ', fruits)

# Basic element-wise operations using arithmetic symbols
print()
print('fruits + 2:\n', fruits + 2) # Add 2 to each item in fruits
print()
print('fruits - 2:\n', fruits - 2) # We subtract 2 to each item in fruits
print()
print('fruits *2:\n', fruits* 2) # Multiply each item in fruits by 2
print()
print('fruits / 2:\n', fruits / 2) # Divide each item in fruits by 2
```

# *Creating Pandas DataFrames
Create adictionary that contains items purchased by two people, Alice and Bob, on an online store. The Pandas Series will use the priceof the items purchased as data,and the purchased items will be used as the index labels to the Pandas Series.

```python
# We create Pandas as pd into Python
import pandas as pd

# We Create a dictionary of Pandas Series
items = {'Alice': pd.Series(data = [40, 110, 500, 45], index = ['book', 'glasses', 'bike', 'pants']),
        'Bob': pd.Series(data = [245, 25, 55], index = ['bike', 'pants', 'watch'])}

# Print the type of items to see that it is a dictionary
print(type(items))
```

```python
# Create a DataFrame using a dictionary of Series

# We create a Pandas DataFrame by passing it a dictionary of Pandas Series
shopping_carts = pd.DataFrame(items)

# We display the DataFrame
print(shopping_carts)
```
