```python
# Count the number of Pokémon in each generation

generation_counts = pokemon['generation_id'].value_counts().sort_index()

# Display the counts

generation_counts
```

```python
# Create the bar chart

sb.barplot(
    x=generation_counts.index,
    y=generation_counts.values
)

# Add title and labels

plt.title('Number of Pokémon by Generation')
plt.xlabel('Generation')
plt.ylabel('Number of Pokémon')

# Display the chart

plt.show()
```

```python
# Create the bar chart

plt.figure(figsize=(10, 6))

sb.barplot(
    x=generation_counts.index,
    y=generation_counts.values
)

# Add values above each bar

for i, count in enumerate(generation_counts.values):
    plt.text(
        i,
        count + 5,
        count,
        ha='center',
        va='bottom'
    )

# Add title and labels

plt.title('Number of Pokémon Introduced by Generation')
plt.xlabel('Generation')
plt.ylabel('Number of Pokémon')

# Improve layout

plt.tight_layout()

# Display the chart

plt.show()
```

```python
# Reshape the Pokémon type data

pkmn_types = pokemon.melt(
    id_vars=['id', 'species'],
    value_vars=['type_1', 'type_2'],
    var_name='type_level',
    value_name='type'
).dropna()
```

```python
# Display the first 10 rows

pkmn_types.head(10)
```

```python
# Count the number of Pokémon belonging to each type

type_counts = pkmn_types['type'].value_counts()

# Display the results

type_counts
```

```python
# Create a bar chart of Pokémon type frequency

plt.figure(figsize=(12, 6))

sb.barplot(
    x=type_counts.index,
    y=type_counts.values
)

# Rotate the labels

plt.xticks(rotation=90)

# Add title and labels

plt.title('Frequency of Pokémon Types')
plt.xlabel('Pokémon Type')
plt.ylabel('Number of Pokémon')

# Improve the layout

plt.tight_layout()

# Display the chart

plt.show()
```

```python
# Calculate the relative frequency of each Pokémon type

type_frequency = pkmn_types['type'].value_counts(normalize=True)

# Sort from most frequent to least frequent

type_frequency = type_frequency.sort_values(ascending=False)

# Display the results

type_frequency
```

```python
# Create the relative frequency bar chart

plt.figure(figsize=(12, 6))

sb.barplot(
    x=type_frequency.index,
    y=type_frequency.values
)

# Rotate the x-axis labels

plt.xticks(rotation=90)

# Add title and labels

plt.title('Proportion of Pokémon by Type')
plt.xlabel('Pokémon Type')
plt.ylabel('Proportion')

# Improve the layout

plt.tight_layout()

# Display the chart

plt.show()
```

```python
# Convert relative frequencies to percentages

type_percentage = (
    pkmn_types['type']
    .value_counts(normalize=True)
    .sort_values(ascending=False) * 100
)

# Display the percentages

type_percentage
```

```python
# Create the chart

plt.figure(figsize=(12, 6))

sb.barplot(
    x=type_percentage.index,
    y=type_percentage.values
)

# Add percentage values above each bar

for i, percentage in enumerate(type_percentage.values):
    plt.text(
        i,
        percentage + 0.3,
        f'{percentage:.1f}%',
        ha='center',
        va='bottom'
    )

# Add title and labels

plt.title('Percentage Distribution of Pokémon Types')
plt.xlabel('Pokémon Type')
plt.ylabel('Percentage (%)')

# Rotate x-axis labels

plt.xticks(rotation=90)

# Improve layout

plt.tight_layout()

# Display the chart

plt.show()
```

```python
# Create the percentage bar chart

plt.figure(figsize=(12, 6))

sb.barplot(
    x=type_percentage.index,
    y=type_percentage.values
)

# Rotate the labels

plt.xticks(rotation=90)

# Add title and labels

plt.title('Percentage Distribution of Pokémon Types')
plt.xlabel('Pokémon Type')
plt.ylabel('Percentage (%)')

# Improve the layout

plt.tight_layout()

# Display the chart

plt.show()
```

```python
# Count Pokémon with and without a secondary type

secondary_type_counts = pokemon['type_2'].notna().value_counts()

secondary_type_counts
```

```python
# Create labels for the chart

labels = ['No Secondary Type', 'Has Secondary Type']

# Create the chart

plt.figure(figsize=(8, 5))

sb.barplot(
    x=secondary_type_counts.index,
    y=secondary_type_counts.values
)

# Replace x-axis labels

plt.xticks(
    ticks=[0, 1],
    labels=labels
)

# Add title and labels

plt.title('Pokémon with and without a Secondary Type')
plt.xlabel('Secondary Type Status')
plt.ylabel('Number of Pokémon')

plt.tight_layout()

plt.show()
```
